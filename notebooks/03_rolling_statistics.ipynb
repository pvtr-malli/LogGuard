# Time-Based Statistics Features - Stats per 30s Window Only

Calculate statistics **within each 30-second window** only (no rolling/lookback):
- Mean values per window
- Standard deviation per window  
- Max/Min values per window
- Percentiles (p95, p99) per window

**Note:** We DO NOT compute rolling statistics (5min, 15min, 1h) because:
- In streaming, we only have the current 30s window
- No access to historical windows during real-time processing
- This matches production streaming pipeline behavior

**Why These Matter:**
- Detects spikes within the window
- Captures variability within 30s
- Identifies tail behavior in current window

## Summary

### Design Decision: No Rolling Features

**Why:**
- In streaming, we only have the current 30-second window buffer
- Cannot look back at previous windows (no historical state)
- To match training with production, we use only per-window features

### Features Used (from notebook 02):

**Volume Features (per 30s window):**
- `log_count`: Total logs in window
- `error_count`, `fatal_count`, `warn_count`: Counts by level
- `error_rate`, `fatal_rate`, `critical_rate`: Ratios
- `warn_to_info_ratio`: Warning to info ratio

**These are sufficient because:**
1. Anomalies manifest within 30s windows
2. DBSCAN can detect patterns across windows spatially
3. Text features (next notebook) add semantic information
4. Simpler model = easier to deploy and debug
